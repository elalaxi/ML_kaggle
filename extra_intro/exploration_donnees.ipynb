```python
#importations

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy as sp
import sklearn as sk

#print("Tous les modules sont prêts à être utilisés !")

```

```python
df = pd.read_csv("extra_intro.csv", delimiter=",")
df.head()

print("Number of rows :", df.shape[0])
print("Number of columns :", df.shape[1])
df.info()
df.describe()

for col in df.columns:
    print(f"\n--- {col} ---")
    print(df[col].unique())
```

```python
# Valeurs manquantes ?

missing = df.isna().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print(missing)

# aucun missing
```

```python
# Distribution de la variable cible

plt.figure(figsize=(7, 5))
sns.countplot(data=df,x="Personality")

plt.title("Distribution of personalities")
plt.xlabel("Personality")
plt.ylabel("Number of individuals")

plt.show()

df["Personality"].value_counts(normalize=True) * 100
```

On voit que les classes sont plutôt équilibrées en terme d'effectif

```python
numeric_cols = df.select_dtypes(include="number").columns
print(numeric_cols)

df[numeric_cols].hist(figsize=(14, 10),bins=12)

plt.suptitle("Distribution of numeric variables", fontsize=16)
plt.show()
```

```python
for col in numeric_cols:

    plt.figure(figsize=(8, 4))

    sns.histplot(
        data=df,
        x=col,
        kde=True
    )

    plt.title("Distribution of " + col)
    plt.xlabel(col)
    plt.ylabel("Number of individuals")

    plt.show()
```

```python
for col in numeric_cols:

    plt.figure(figsize=(8, 3))

    sns.boxplot(
        data=df,
        x=col
    )

    plt.title(f"Boxplot de {col}")

    plt.show()
```
